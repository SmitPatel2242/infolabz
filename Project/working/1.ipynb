```python
import pandas as pd
```

```python
data = pd.read_csv("ipl_data_cleaned.csv")
data['final_total_runs']=0

total_runs = data.groupby(["id","innings"])["total_runs"].sum().reset_index()
for i, row in total_runs.iterrows():
    # find the corresponding rows in the main dataframe and update the final_total_runs column
    data.loc[(data['id'] == row['id']) & (data['innings'] == row['innings']), 'final_total_runs'] = row['total_runs']
data
```

```python
# Let's group the dataframe by match_id and innings
grouped = data.groupby(["id", "innings"])

# Let's calculate the cumulative sum of wickets, runs, and balls
data["wickets"] = grouped["is_wkt_delivery"].cumsum()
data["runs"] = grouped["total_runs"].cumsum()
# df[['match_id','is_wicket','wickets','total_run', 'runs','over']].head(15)
```

```python
data.to_csv('match_data.csv', index=False)
data = data[data["innings"]==1]
delete = ['id','season','innings','is_super_over','batter','non_striker','bowler','runs_off_bat','runs_from_extras','total_runs','is_wkt_delivery','player_out','wkt_type','fielders_involved','byes','legbyes','wides','noballs','penalty']
data = data.drop(delete,axis=1)
data = data[data["overs_balls"]>=5.0]
```

```python
encode_data = pd.get_dummies(data=data,columns=['batting_team','bowling_team'])
encode_data
```
